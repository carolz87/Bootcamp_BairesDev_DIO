```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Carregar a imagem em escala de cinza
imagem_orig = cv2.imread('Imagens/exemplo.jpg', cv2.IMREAD_COLOR_RGB)
```

```python

# Carregar a imagem em escala de cinza
imagem = cv2.imread('Imagens/exemplo.jpg', cv2.IMREAD_GRAYSCALE)
```

```python

# Verificar se a imagem foi carregada corretamente
if imagem is None:
    print("Erro ao carregar a imagem.")
    exit()
```

```python

# Aplicar a binarização utilizando o método de Otsu
_, imagem_binarizada = cv2.threshold(imagem, 0, 255, cv2.THRESH_BINARY + cv2.THRESH_OTSU)
```

```python

# Exibir a imagem original e a imagem binarizada
plt.figure(figsize=(10, 5))

plt.subplot(1, 3, 1)
plt.title('Imagem Original')
plt.imshow(imagem_orig)
plt.axis('off')

plt.subplot(1, 3, 2)
plt.title('Imagem Cinza')
plt.imshow(imagem, cmap='gray')
plt.axis('off')

plt.subplot(1, 3, 3)
plt.title('Imagem Binarizada')
plt.imshow(imagem_binarizada, cmap='gray')
plt.axis('off')

plt.show()
```
